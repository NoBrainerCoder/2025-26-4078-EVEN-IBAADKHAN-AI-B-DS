```python
#Download and save the dataset in the local drive
import pandas as pd
#url of dataset(sample csv dataset)
url="https://raw.githubusercontent.com/mwaskom/seaborn-data/refs/heads/master/iris.csv"
#read dataset from url
df=pd.read_csv(url)
#display first few rows
print(df.head())
#save dataset to local drive
df.to_csv("iris.csv",index=False)
print("dataset saved successfully")
```

```python
#List
superstore=["metro",'tesco','walmart','kmart','carrefour']
print(superstore)
number_list=[4,4,6,7,2,9,10,15]
mixed_list=['cat',87.23,65,[9,7,3,3]]
print(type(mixed_list))
#list operations
list1=[3,5,7,1]
list2=[2,4,6,8]
print(list1+list2) #list concartination
print(list1*3)
print(list1==list2)
#string to list
quote='Hello World'
string_to_list=list(quote)
print(string_to_list)
# list indexing, slicing
print(list1[:2])
print(list1[-2:])
print(5 in list1)
print(11 in list1)
#list built-in functions
cars=['BMW','Audi','Mercedes','Toyota']
print(len(cars))
print(max(number_list))
print(min(number_list))
print(sum(number_list))
print(sorted(list1))
print(sorted(list1,reverse=True))
#list methods
cars.append('Tesla')
print(cars)
cars.insert(2,'Ford')
print(cars)
cars.remove('Tesla')
print(cars)
cars.pop()
print(cars)
#Nested list
nested_list=[[1,2,3],[4,5,6],[7,8,9]]
print(nested_list)
print(nested_list[1])
print(nested_list[1][2])
```

```python
#Creating tuples
t1 = (10, 20, 30, 40)
t2 = (50, 60)

# Accessing elements
print(t1[0])
print(t1[-1])

# Slicing
print(t1[1:3])
print(t1[::-1])

# Concatenation
t3 = t1 + t2
print(t3)

# Repetition
t4 = t1 * 2
print(t4)

# Membership
print(20 in t1)
print(100 not in t1)

# Length
print(len(t1))

# Iteration
for i in t1:
    print(i)

# Tuple methods
print(t1.count(20))
print(t1.index(30))
```
